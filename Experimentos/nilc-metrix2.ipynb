```python
import pandas as pd
import requests

URL = "http://localhost:8081/api/v1/metrix/_min/yyy?format=json"
df = pd.read_csv("df_papers_caixa_2.csv").head(4)
resultados = []

for i, row in df.iterrows():
    texto = str(row['Abstract'])
    print(f"Enviando Texto {i+1} para a porta 8081...")
    resp = requests.post(URL, data=texto.encode('utf-8'), headers={'Content-Type': 'text/plain'}, timeout=None)
    resultados.append(resp.json() if resp.status_code == 200 else {"error": resp.status_code})

pd.DataFrame(resultados).to_csv("resultado_final_2.csv", index=False)
```
