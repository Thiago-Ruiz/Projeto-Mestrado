```python
import pandas as pd
import os
```

```python
# Criar diretório para cada df
os.makedirs("main_txt", exist_ok=True)
os.makedirs("geral_txt", exist_ok=True)
```

```python
def convert_csv_to_txt(index, abstract):
    # Criar o nome do arquivo txt com base no ID
    nome_arquivo = f"geral_txt/{index}.txt"
    
    # Escrever o conteúdo do abstract no arquivo txt
    with open(nome_arquivo, 'w', encoding='utf-8') as file:
        file.write(abstract)
    
    return nome_arquivo
    
```

```python
main_df = pd.read_csv('Dataset_Textos_Hum_IA.csv')
Geral_df = pd.read_csv('Geral.csv')
```

```python
main_df.info()
```

```python
for index, row in main_df.iterrows():
    
    arquivo_txt = convert_csv_to_txt(row['Index'], row['Abstract'])
    
    # Armazenar no diretório main_txt
    os.rename(arquivo_txt, f"main_txt/{row['Index']}.txt")
```

```python
for index, row in Geral_df.iterrows():
    
    arquivo_txt = convert_csv_to_txt(row['index'], row['Abstract'])
    
    # Armazenar no diretório main_txt
    os.rename(arquivo_txt, f"geral_txt/{row['index']}.txt")
```
