```python
import pandas as pd
import requests 
from bs4 import BeautifulSoup
```

Springer Nature

```python
df = pd.DataFrame(columns=['url'])

for page in range(1, 8):
    req = requests.get(f"https://link.springer.com/search?query=%28%22AI-generated+text%22+OR+%22LLM-generated+text%22+OR+%22ChatGPT-generated+text%22%29+AND+%28detection+OR+classifier+OR+authorship%29+AND+%28%22academic+writing%22+OR+%22scientific+writing%22%29&content-type=Article&content-type=Research&content-type=Conference+Paper&content-type=Review&content-type=Reference+Work+Entry&content-type=News&date=custom&dateFrom=2020&dateTo=2025&sortBy=relevance&page={page}")
    if  req.status_code == 200:
        print("Conexão bem sucedida! Extraindo dados da página", page)
        print("\n")
    soup = BeautifulSoup(req.content, 'html.parser')
    #print(soup.prettify())
    
    list_url = soup.find_all(class_="app-card-open__link")
    #print(list_url)
    
    for url in list_url:
        print(url.get('href'))
        df = pd.concat([df, pd.DataFrame({'url':[url.get('href')]})], ignore_index=True)

df['url'] = "https://link.springer.com" + df['url']
df.to_csv('df_artigos_springer_string_5.csv', index=True)
```
