#### Segmentação do Dataframe por Classe

```python
# Classe Humana

def calculate_metrics(df, class_name, start_col=4):

    metric_cols = df.columns[start_col:]

    df[metric_cols] = df[metric_cols].apply(
        pd.to_numeric,
        errors='coerce'
    )

    metrics = []

    for column in metric_cols:

        media = df[column].mean()
        mediana = df[column].median()
        desvio_padrao = df[column].std()

        cv = (
            (desvio_padrao / media) * 100
            if media != 0 and not np.isnan(media)
            else np.nan
        )

        metrics.append({
            'Classe': class_name,
            'Métrica': column,
            'Média': media,
            'Mediana': mediana,
            'Desvio Padrão': desvio_padrao,
            'Coeficiente de Variação (%)': cv
        })

    return pd.DataFrame(metrics)
```

```python
df_human_metrics = calculate_metrics(df_main_human_text, 'Humana')
df_human_metrics.to_csv('human_metrics.csv', index=False)

df_polished_metrics = calculate_metrics(df_main_polished_ai_text, 'Polida_IA')
df_polished_metrics.to_csv('polished_metrics.csv', index=False)

df_generated_metrics = calculate_metrics(df_main_generated_ai_text, 'Gerada')
df_generated_metrics.to_csv('generated_metrics.csv', index=False)
```
